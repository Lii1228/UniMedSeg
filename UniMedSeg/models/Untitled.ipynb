# Inspect the released networks

Run this notebook from the repository or its models directory. Select a variant to inspect the backbone and run a 2D visual-context forward pass.


```python
from pathlib import Path
import sys

import torch

root = next(p for p in (Path.cwd(), *Path.cwd().parents) if (p / "run_training.sh").is_file())
sys.path.insert(0, str(root))

from UniMedSeg.models.backbones import MODEL_CONFIGS, build_primus

for name, (width, depth, heads) in MODEL_CONFIGS.items():
    print(f"UniMedSeg-{name}: width={width}, blocks={depth}, heads={heads}")
```

```python
variant = "S"
model = build_primus(variant).eval()
image = torch.rand(1, 1, 128, 128)
support = torch.rand(1, 1, 1, 128, 128)
mask = (support > 0.5).float()

with torch.inference_mode():
    prediction = model(image, support, mask)
assert prediction.shape == image.shape
print(f"Output shape: {tuple(prediction.shape)}")
```
